```python
import pandas as pd, glob

frames = []
for f in sorted(glob.glob("/content/Skateboard_injuries/neiss/neiss*.tsv")):
    df = pd.read_csv(f, sep="\t")
    frames.append(df)

neiss = pd.concat(frames)
print(neiss.head())
```

```python
neiss["Product_1"] = pd.to_numeric(neiss["Product_1"], errors="coerce")
neiss["Product_2"] = pd.to_numeric(neiss["Product_2"], errors="coerce")

skate = neiss[(neiss["Product_1"] == 1333) | (neiss["Product_2"] == 1333)].copy()
print(len(skate))
```

```python
print(skate)
```

```python
skate.columns = skate.columns.str.strip().str.lower()
skate.columns
```

```python
skate.to_csv("neiss_skateboard.csv", index=False)
print(skate.shape)
print(skate.isnull().mean().round(2))
print(skate.dtypes)
```
